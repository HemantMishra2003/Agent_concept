```python
from typing import TypedDict
from langgraph.graph import StateGraph, START, END
```

```python
class State(TypedDict):
    name: str         # State naam ki class/type definition banayi
                      # Uske andar name naam ki ek key define ki
                      # name ki value ka type str hona chahiye
```

```python
def hello_node(state): # state is coming here , means here state  is an input
    return {
        "name": "Hello " + state["name"] , # now state will be updated here
    }                                      # name : hello hemant

    #  here  basic things is happening that we take first state input
    # put into another node of state input.

#  The same workflow State keeps getting updated,
#   and the next node can receive the updated State.
```

```python
graph = StateGraph(State) # making  garph for state

graph.add_node("hello", hello_node) # now we are adding  hello node into the graph
                                    # here hello is just name of
                                    # hello_node function
graph.add_edge(START, "hello") # this  is edges and this will guide node
                               # to which direction u have to run
graph.add_edge("hello", END)   #
```

```python
app = graph.compile()   # Compile
```

```python
app.invoke({"name": "shivam bhaiya"})
```

 we have till yet made 1 node , we will connect second node





```python
def bye_node(state): # making 2 nd node bye_node
    return {
        "name": state["name"] + " Bye"
    }
```

```python
graph.add_node("bye", bye_node)
```

```python
graph.add_edge("hello", "bye")
```

```python
graph.add_edge("bye", END)
```

```python
app = graph.compile()
```

```python
result = app.invoke({"name": "Shivam"})
```
